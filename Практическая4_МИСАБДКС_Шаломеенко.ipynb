Задача 1. Вектор из 10 нулей

```python
import numpy as np

v = np.zeros(10)
print(v)
```

Задача 2. Вектор из 10 единиц

```python
v = np.ones(10)
print(v)
```

Задача 3. Вектор из 10 элементов, заполненный 2.5

```python
v = np.full(10, 2.5)
print(v)
```

Задача 4. Вектор из нулей, пятый элемент = 1

```python
v = np.zeros(10)
v[4] = 1
print(v)
```

Задача 5. Вектор со значениями от 10 до 49

```python
v = np.arange(10, 50)
print(v)
```

Задача 6. Развернуть вектор

```python
v = np.arange(10)
print("Исходный:", v)

reversed_v = v[::-1]
print("Развёрнутый:", reversed_v)
```

Задача 8. Индексы ненулевых элементов

Задача 9. Массив 3×3×3 со случайными значениями

Задача 10. Массив 10×10 со случайными значениями; min и max

```python
arr = np.random.rand(10, 10)
print("Минимум:", arr.min())
print("Максимум:", arr.max())
```

Задача 11. Матрица с 0 внутри и 1 по границам

```python
n = 10
m = np.ones((n, n), dtype=int)
m[1:-1, 1:-1] = 0
print(m)
```

Задача 12. Шахматная доска 8×8

```python
board = np.zeros((8, 8), dtype=int)
board[1::2, ::2] = 1
board[::2, 1::2] = 1
print(board)
```

Задача 13. Перемножить матрицы 5×3 и 3×2

```python
A = np.random.randint(0, 10, size=(5, 3))
B = np.random.randint(0, 10, size=(3, 2))

C = A @ B          # или A.dot(B), или np.matmul(A, B)
print("A:", A.shape)
print("B:", B.shape)
print("C (5×2):", C.shape)
print(C)
```

Задача 14. Матрица 5×7 из целых 1..100; удалить 3-й и 5-й столбцы

```python
arr = np.random.randint(1, 101, size=(5, 7))
print("Исходная:\n", arr)
print("Форма:", arr.shape)

new_arr = np.delete(arr, [3, 5], axis=1)
print("После удаления:\n", new_arr)
print("Новая форма:", new_arr.shape)
```

Задача 15. Массив из генератора

```python
def generate():
    for x in range(10):
        yield x

arr = np.fromiter(generate(), dtype=int)
print(arr)
```

Задача 16. Двумерный список → NumPy-массив (int32)

```python
nested = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

arr = np.array(nested, dtype=np.int32)
print(arr)
print("Тип:", arr.dtype)
print("Форма:", arr.shape)
```

Задача 17. Проверить, одинаковы ли два NumPy-массива

```python
a = np.array([1, 2, 3])
b = np.array([1, 2, 3])
c = np.array([1, 2, 4])

print(np.array_equal(a, b))
print(np.array_equal(a, c))
```

Задача 18. Сделать массив неизменяемым

```python
a = np.array([1, 2, 3])
a.flags.writeable = False

print("Writeable:", a.flags.writeable)

try:
    a[0] = 100
except ValueError as e:
    print("Ошибка:", e)
```
