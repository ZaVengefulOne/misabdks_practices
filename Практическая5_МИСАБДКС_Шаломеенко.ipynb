# Импорты и загрузка данных

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.dpi"] = 110

# Загружаем данные (разделитель ';', первая строка — заголовки на англ.,
# вторая — русские, она будет отброшена
female = pd.read_csv("female_names.csv", sep=";", skiprows=1)
male   = pd.read_csv("male_names.csv",   sep=";", skiprows=1)

female = female[female["Code"] != "Код"].copy()
male   = male[male["Code"]   != "Код"].copy()

for df in (female, male):
    df["NumberOfPersons"] = pd.to_numeric(df["NumberOfPersons"], errors="coerce")
    df["Year"]            = pd.to_numeric(df["Year"], errors="coerce")
    df["Name"]            = df["Name"].astype(str).str.strip()
    df["Month"]           = df["Month"].astype(str).str.strip().str.lower()

month_map = {
    "январь":"Январь", "февраль":"Февраль", "март":"Март", "апрель":"Апрель",
    "май":"Май", "июнь":"Июнь", "июль":"Июль", "август":"Август",
    "сентябрь":"Сентябрь", "октябрь":"Октябрь", "ноябрь":"Ноябрь", "декабрь":"Декабрь",
}
female["Month"] = female["Month"].map(month_map)
male["Month"]   = male["Month"].map(month_map)

female["Gender"] = "Ж"
male["Gender"]   = "М"

MONTHS = ["Январь","Февраль","Март","Апрель","Май","Июнь",
          "Июль","Август","Сентябрь","Октябрь","Ноябрь","Декабрь"]

female_p = female[(female["Year"] >= 2015) & (female["Year"] <= 2025)].copy()
male_p   = male[(male["Year"]   >= 2015) & (male["Year"]   <= 2025)].copy()

print("Female:", female.shape, "| Male:", male.shape)
print("Период 2015-2025 -> Female:", female_p.shape, "| Male:", male_p.shape)
female.head()
```

Задача 1. Топ-100 популярных имён за 2015–2025

```python
both_p = pd.concat([female_p, male_p], ignore_index=True)

top100 = (
    both_p.groupby("Name", as_index=False)["NumberOfPersons"]
          .sum()
          .sort_values("NumberOfPersons", ascending=False)
          .head(100)
)

plt.figure(figsize=(12, 22))
sns.barplot(data=top100, y="Name", x="NumberOfPersons", palette="viridis")
plt.title("Топ-100 популярных имён (2015–2025)", fontsize=14)
plt.xlabel("Суммарное число новорождённых")
plt.ylabel("Имя")
plt.tight_layout()
plt.show()

top100.head(10)
```

Задача 2. Топ-10 мужских имён за каждый месяц

```python
fig, axes = plt.subplots(6, 2, figsize=(14, 26))
axes = axes.flatten()

for i, month in enumerate(MONTHS):
    m = male_p[male_p["Month"] == month]
    top = (
        m.groupby("Name", as_index=False)["NumberOfPersons"]
         .sum()
         .sort_values("NumberOfPersons", ascending=False)
         .head(10)
    )
    sns.barplot(data=top, y="Name", x="NumberOfPersons", ax=axes[i], palette="Blues_r")
    axes[i].set_title(f"Топ-10 мужских имён — {month}", fontsize=11)
    axes[i].set_xlabel("Число новорождённых")
    axes[i].set_ylabel("")

plt.suptitle("Топ-10 мужских имён по месяцам (агрегация за 2015–2025)", fontsize=15, y=1.005)
plt.tight_layout()
plt.show()
```

Задача 3. Топ-10 женских имён за каждый месяц

```python
fig, axes = plt.subplots(6, 2, figsize=(14, 26))
axes = axes.flatten()

for i, month in enumerate(MONTHS):
    f = female_p[female_p["Month"] == month]
    top = (
        f.groupby("Name", as_index=False)["NumberOfPersons"]
         .sum()
         .sort_values("NumberOfPersons", ascending=False)
         .head(10)
    )
    sns.barplot(data=top, y="Name", x="NumberOfPersons", ax=axes[i], palette="Reds_r")
    axes[i].set_title(f"Топ-10 женских имён — {month}", fontsize=11)
    axes[i].set_xlabel("Число новорождённых")
    axes[i].set_ylabel("")

plt.suptitle("Топ-10 женских имён по месяцам (агрегация за 2015–2025)", fontsize=15, y=1.005)
plt.tight_layout()
plt.show()
```

Задача 4. Топ-10 мужских имён за весь период

```python
top10_m = (
    male_p.groupby("Name", as_index=False)["NumberOfPersons"]
          .sum()
          .sort_values("NumberOfPersons", ascending=False)
          .head(10)
)

plt.figure(figsize=(10, 6))
sns.barplot(data=top10_m, y="Name", x="NumberOfPersons", palette="Blues_r")
plt.title("Топ-10 мужских имён за 2015–2025")
plt.xlabel("Суммарное число новорождённых")
plt.ylabel("Имя")
plt.tight_layout()
plt.show()

top10_m
```

Задача 5. Топ-10 женских имён за весь период

```python
top10_f = (
    female_p.groupby("Name", as_index=False)["NumberOfPersons"]
            .sum()
            .sort_values("NumberOfPersons", ascending=False)
            .head(10)
)

plt.figure(figsize=(10, 6))
sns.barplot(data=top10_f, y="Name", x="NumberOfPersons", palette="Reds_r")
plt.title("Топ-10 женских имён за 2015–2025")
plt.xlabel("Суммарное число новорождённых")
plt.ylabel("Имя")
plt.tight_layout()
plt.show()

top10_f
```
