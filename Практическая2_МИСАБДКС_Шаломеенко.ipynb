# JSON

Задача 1. Создание JSON-объекта

```python
import json

student = {
    "name": "Иван",
    "age": 20,
    "grades": [5, 4, 5, 3, 4]
}

json_string = json.dumps(student, ensure_ascii=False, indent=4)
print(json_string)
```

Задача 2. Чтение JSON-файла

```python
import json

data = {
    "name": "Иван",
    "age": 20,
    "courses": ["Математика", "Физика"]
}
with open('data.json', 'w', encoding='utf-8') as f:
    json.dump(data, f, ensure_ascii=False, indent=4)

with open('data.json', 'r', encoding='utf-8') as f:
    loaded = json.load(f)
print("Имя студента:", loaded['name'])
```

Задача 3. Модификация данных (добавить email и обновить файл)

```python
import json

with open('data.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

data['email'] = 'ivan@example.com'


with open('data.json', 'w', encoding='utf-8') as f:
    json.dump(data, f, ensure_ascii=False, indent=4)

with open('data.json', 'r', encoding='utf-8') as f:
    print(json.load(f))
```

Задача 4. Массив книг + поиск по автору

```python
import json

books = [
    {"title": "Война и мир", "author": "Толстой", "year": 1869},
    {"title": "Анна Каренина", "author": "Толстой", "year": 1877},
    {"title": "Конец Детства", "author": "Кларк", "year": 1953},
    {"title": "Трудно быть Богом", "author": "Стругацкие", "year": 1964},
    {"title": "Преступление и наказание", "author": "Достоевский", "year": 1866},
]

with open('books.json', 'w', encoding='utf-8') as f:
    json.dump(books, f, ensure_ascii=False, indent=4)


def find_by_author(filename, author):
    with open(filename, 'r', encoding='utf-8') as f:
        data = json.load(f)
    return [book for book in data if book['author'].lower() == author.lower()]


result = find_by_author('books.json', 'Толстой')
for b in result:
    print(b)
```

Задача 5. Обработка ошибок при чтении JSON

```python
import json

def safe_read_json(filename):
    try:
        with open(filename, 'r', encoding='utf-8') as f:
            return json.load(f)
    except FileNotFoundError:
        print(f"Ошибка: файл '{filename}' не найден.")
    except json.JSONDecodeError as e:
        print(f"Ошибка: некорректный JSON в файле '{filename}'. Детали: {e}")
    except Exception as e:
        print(f"Непредвиденная ошибка: {e}")
    return None


print(safe_read_json('data.json'))
print(safe_read_json('nope.json'))
```

Задача 7. Вложенный каталог товаров + поиск

```python
import json

catalog = {
    "Электроника": {
        "Смартфоны": [
            {"name": "iPhone 15", "price": 89990},
            {"name": "Samsung S24", "price": 79990}
        ],
        "Ноутбуки": [
            {"name": "MacBook Air", "price": 119990},
            {"name": "Lenovo IdeaPad", "price": 54990}
        ]
    },
    "Одежда": {
        "Куртки": [
            {"name": "Пуховик Winter", "price": 12000}
        ],
        "Футболки": [
            {"name": "Basic Tee", "price": 1500}
        ]
    }
}

with open('catalog.json', 'w', encoding='utf-8') as f:
    json.dump(catalog, f, ensure_ascii=False, indent=4)


def find_product(catalog, name):
    """Ищет товар по названию, возвращает путь (категория, подкатегория) и данные."""
    name_lower = name.lower()
    for category, subcats in catalog.items():
        for subcat, products in subcats.items():
            for product in products:
                if product['name'].lower() == name_lower:
                    return category, subcat, product
    return None


# Проверка
result = find_product(catalog, "macbook air")
print(result)

print(find_product(catalog, "несуществующий товар"))
```

# CSV

Задача 1. Чтение CSV с оценками студентов

```python
import csv

with open('students.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['Имя', 'Оценка'])
    writer.writerow(['Анна', 5])
    writer.writerow(['Борис', 4])
    writer.writerow(['Виктор', 5])

with open('students.csv', 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

Задача 2. Запись списка словарей (книги) в CSV

```python
import csv

books = [
    {'title': 'Война и мир', 'author': 'Толстой', 'year': 1869},
    {'title': 'Конец детства', 'author': 'Кларк', 'year': 1953}
]

fieldnames = books[0].keys()

with open('books.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(books)

with open('books.csv', 'r', encoding='utf-8') as f:
    print(f.read())
```

Задача 3. Фильтрация продаж (сумма > 1000)

```python
import csv

sales = [
    ['Дата', 'Товар', 'Сумма'],
    ['01.01.2023', 'Телефон', 1200],
    ['02.01.2023', 'Ручка', 100],
    ['03.01.2023', 'Ноутбук', 15000],
]
with open('sales.csv', 'w', newline='', encoding='utf-8') as f:
    csv.writer(f).writerows(sales)

with open('sales.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    filtered = [row for row in reader if float(row['Сумма']) > 1000]

with open('sales_filtered.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['Дата', 'Товар', 'Сумма'])
    writer.writeheader()
    writer.writerows(filtered)

for row in filtered:
    print(row)
```

Задача 4. Средняя оценка по предмету

```python
import csv
import pandas as pd

grades = [
    ['Студент', 'Математика', 'Физика', 'Химия'],
    ['Анна', 5, 4, 5],
    ['Борис', 4, 3, 4],
    ['Виктор', 5, 5, 5],
]
with open('grades.csv', 'w', newline='', encoding='utf-8') as f:
    csv.writer(f).writerows(grades)

with open('grades.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    subjects = [c for c in reader.fieldnames if c != 'Студент']

    totals = {s: 0 for s in subjects}
    counts = {s: 0 for s in subjects}

    for row in reader:
        for s in subjects:
            totals[s] += int(row[s])
            counts[s] += 1

df = pd.read_csv('grades.csv')
print(df[['Математика', 'Физика', 'Химия']].mean())
```

Задача 5. Объединение нескольких CSV с одинаковыми заголовками

```python
import csv

with open('file1.csv', 'w', newline='', encoding='utf-8') as f:
    w = csv.writer(f)
    w.writerow(['ID', 'Имя', 'Возраст'])
    w.writerow([1, 'Анна', 20])
    w.writerow([2, 'Борис', 22])

with open('file2.csv', 'w', newline='', encoding='utf-8') as f:
    w = csv.writer(f)
    w.writerow(['ID', 'Имя', 'Возраст'])
    w.writerow([3, 'Виктор', 21])
    w.writerow([4, 'Галина', 19])


def merge_csv(input_files, output_file):
    header_written = False
    with open(output_file, 'w', newline='', encoding='utf-8') as out:
        writer = csv.writer(out)
        for filename in input_files:
            with open(filename, 'r', encoding='utf-8') as f:
                reader = csv.reader(f)
                header = next(reader)
                if not header_written:
                    writer.writerow(header)
                    header_written = True
                writer.writerows(reader)


merge_csv(['file1.csv', 'file2.csv'], 'merged.csv')

with open('merged.csv', 'r', encoding='utf-8') as f:
    print(f.read())
```
